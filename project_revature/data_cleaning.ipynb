```python
import pandas as pd

# ---------- READ FILES ----------
# season is text because some seasons may look like 2020/21
matches = pd.read_csv(
    "matches.csv",
    dtype={"season": "string"},
    low_memory=False
)

deliveries = pd.read_csv(
    "deliveries.csv",
    dtype={"season": "string"},
    low_memory=False
)

# ---------- CLEAN MATCHES ----------
# Remove duplicate rows
matches = matches.drop_duplicates()

# Remove extra spaces and replace missing text with blank
for col in matches.select_dtypes(include=["object", "string"]).columns:
    matches[col] = matches[col].fillna("").str.strip()

# Replace missing city names
matches["city"] = matches["city"].replace("", "Unknown")

# Convert date column
matches["date"] = pd.to_datetime(matches["date"], errors="coerce")

# ---------- CLEAN DELIVERIES ----------
# Remove duplicate rows
deliveries = deliveries.drop_duplicates()

# Remove extra spaces and replace missing text with blank
for col in deliveries.select_dtypes(include=["object", "string"]).columns:
    deliveries[col] = deliveries[col].fillna("").str.strip()

# Convert numeric columns and replace missing values with 0
number_columns = [
    "innings",
    "over_number",
    "ball",
    "runs_off_bat",
    "extras",
    "total_runs",
    "wides",
    "noballs",
    "byes",
    "legbyes",
    "penalty"
]

for col in number_columns:
    deliveries[col] = pd.to_numeric(
        deliveries[col],
        errors="coerce"
    ).fillna(0).astype(int)

# Convert date column
deliveries["date"] = pd.to_datetime(deliveries["date"], errors="coerce")

# ---------- SAVE CLEANED FILES ----------
matches.to_csv("matches_cleaned.csv", index=False)
deliveries.to_csv("deliveries_cleaned.csv", index=False)

# ---------- CONFIRM RESULT ----------
print("Cleaning completed successfully!")
print("Matches shape:", matches.shape)
print("Deliveries shape:", deliveries.shape)
print("\nNew files created:")
print("1. matches_cleaned.csv")
print("2. deliveries_cleaned.csv")
```

```python
# Check for missing values
print("MISSING VALUES IN MATCHES:")
print(matches.isnull().sum())

print("\nMISSING VALUES IN DELIVERIES:")
print(deliveries.isnull().sum())

# Preview cleaned data
print("\nMATCHES DATA:")
display(matches.head())

print("\nDELIVERIES DATA:")
display(deliveries.head())

# Check duplicate rows
print("\nDuplicate rows in matches:", matches.duplicated().sum())
print("Duplicate rows in deliveries:", deliveries.duplicated().sum())
```
